```python
#import libraries 

import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib 
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure
%matplotlib inline
matplotlib.rcParams['figure.figsize']=(12,8)

#read in the data 

df=pd.read_csv(r'C:\Users\user\Downloads\movies.csv')
```

```python
#looking at the data 

df.head()
```

```python
#let's use if there is any misssing data 

for col in df.columns:
      pct_missing= np.mean(df[col].isnull())
      print('{}-{}%'.format(col, pct_missing))
```

```python
#data types for our columns 
df.dtypes
```

```python
#changing data type of columns 

df['budget'] = df['budget'].fillna(0).astype('int64')
df['gross'] = df['gross'].fillna(0).astype('int64')
```

```python
df.head()
```

```python
#creating correct year column 

df['yearcorrect']=df['released'].astype(str).str[:4]
df.head()
```

```python
df.sort_values(by=['gross'], inplace=False, ascending=False)
```

```python
# If df is a DataFrame and you want to sort by the 'gross' column
df.sort_values(by='gross', ascending=False).head()

# OR if df['gross'] is a Series
df['gross'].sort_values(ascending=False).head()

# OR if df itself is a Series
df.sort_values(by='gross', ascending=False).head()
```

```python
#drop any duplicates

df['company'].drop_duplicates().sort_values(ascending=False)
```

```python
#scatter plot with budget vs gross 

plt.scatter(x=df['budget'], y=df['gross'])

plt.title('budget vs gross earning')
plt.xlabel('gross Earnings')
plt.ylabel('Budget for film')   
plt.show()
```
